```python
-- Retrieve all Customers
-- Usage of WHERE
%%sql
SELECT * FROM customers;
```

```python
%%sql
-- Usage of SELECT
-- Get customer's name, country and score
SELECT first_name,
       country,
       score
FROM customers;
```

```python
%%sql
-- Usage of WHERE
-- Retrive customers with score not equal to 0

SELECT * 
from customers
where score != 0;
```

```python
%%sql
-- Usage of WHERE
-- Retrieve customers from Germany
SELECT *
FROM customers
WHERE country = 'Germany';
```

```python
%%sql
-- Usage of Order By
-- Retrieve all customers and sort the results by the highest score first

SELECT *
FROM Customers
ORDER BY Score desc;
```

```python
%%sql
-- Usage of Order By
-- Retrieve all customers and sort the results by the lowest score first

SELECT *
FROM Customers
ORDER BY Score;
```

```python
%%sql
-- Usage of Nested Order By
-- Retrieve all customers and sort the results by the country and then highest score(Nested Order By)

SELECT *
FROM Customers
ORDER BY country asc, score desc;
```

```python
%%sql
-- Usage of Group By
-- Find the total score for each country

SELECT country, SUM(Score) as score
FROM customers 
Group By country 
```

```python
%%sql
-- Usage of Group By
-- Find the total score and total number of customers for each country

SELECT 
        country,
        SUM(Score) as Score,
        COUNT(Country) as Country
FROM CUSTOMERS
GROUP BY country
```

```python
%%sql
-- Find the averge score for each country considering only customers with score not equal to 0 
-- and return only those countries with an avergae score greater than 430

-- PART 1
SELECT 
    country,
    AVG(Score)
FROM customers
where score != 0
group by country
```

```python
%%sql
-- Usage of Group By
-- 1.Find the avg score for each country considering only customers with a score not equal to zero
-- and return only those countries with an average score greater than 430

--PART 1
SELECT country,
        avg(score)
from Customers
where score != 0
group by country
```

```python
%%sql
-- PART 2
SELECT country,
     avg(score) as Average_Score
from Customers
where score != 0
group by country
having avg(score) > 430
```

```python
%%sql
-- Usage of Distinct(Removes Duplicates)
-- Return Unique list of all countries

SELECT DISTINCT country
FROM CUSTOMERS

```

```python
%%sql
-- Usage of TOP(Restrict the number of Rows Returned).
-- Retrieve only 3 customers with highest score
-- Note : SQLite does not TOP, using LIMIT

SELECT *
FROM customers
order by score desc
LIMIT 3;
```

```python
%%sql
-- Usage of TOP(Restrict the number of Rows Returned).
-- Retrieve only 3 customers with lowest score
-- Note : SQLite does not TOP, using LIMIT
SELECT *
FROM customers
order by score 
LIMIT 3;
```

```python
%%sql

-- Retrieve the two most recent orders

SELECT * 
FROM Orders
Order by order_date desc
limit 2;
```
