```python
%%sql 
-- Inner Join (Only returns matching rows from both tables)
-- Get all customers along with their orders, but only for customers who have placed an order

SELECT 
    c.id,
    o.order_id,
    o.sales
FROM CUSTOMERS as c
INNER JOIN ORDERS as o
ON c.id = o.customer_id
```

```python
# Left Join - Returns all the table from left and only the matching from rightb
```

```python
%%sql
-- Get all customers along with their orders, including those without orders

select c.id,
        c.first_name,
       o.order_id,
       o.sales
from customers c
left join orders o
on c.id=o.customer_id
```

```python
# Right Join - Returns all rows from right and only matching from the left
```

```python
%%sql 
-- Get all customers along with their orders, including orders without matching customers

SELECT c.id,
       c.first_name,
       o.order_id,
       o.sales 
FROM customers c
RIGHT JOIN orders o
ON c.id = o.customer_id
```

```python
%%sql 
-- Get all customers along with their orders, including orders 
-- without matching customers(using left join)

SELECT c.id,
       c.first_name,
       o.order_id,
       o.sales 
FROM orders o
left JOIN customers c
ON o.customer_id = c.id;
```

```python
#FULL join(Return all joins from both the tables- retrun everything)
```

```python
%%sql 
-- Get all customers and all orders even if there's no match

SELECT c.id,
       c.first_name,
       o.order_id,
       o.sales 
FROM orders o
FULL JOIN customers c
ON o.customer_id = c.id;
```

```python
# Advanced SQL JOINS
# Left Anti Join - Return row from left that has no match in right(right is just a return)
```

```python
%%sql
-- Get all customers who haven't placed any order
SELECT c.id,
       c.first_name,
       o.order_id,
       o.sales 
FROM customers c
left JOIN orders o
ON c.id = o.customer_id
WHERE o.customer_id is NULL;
```

```python
# Right Anti Join - Return row from right that has no match in left(left is just a filter)
```

```python
%%sql
-- Get all orders without matching customers
SELECT c.id,
       c.first_name,
       o.order_id,
       o.sales 
FROM customers c
RIGHT JOIN orders o
ON c.id = o.customer_id
where c.id is NULL;
```

```python
%%sql
-- Get all orders without matching customers(using left join)
SELECT c.id,
       c.first_name,
       o.order_id,
       o.sales 
FROM orders o
LEFT JOIN customers c 
ON o.customer_id = c.id
where c.id is NULL;
```

```python
# Full anti join- Return only rows that dont match in either tables
```

```python
%%sql 
-- Get all customers without orders and all orders without customers

SELECT c.id,
       c.first_name,
       o.order_id,
       o.sales 
FROM orders o
FULL JOIN customers c
ON o.customer_id = c.id
where o.customer_id is NULL or c.id is NULL;
```

```python
# CROSS JOIN - Combines every row from left with every row from right 
# All possible combinations - Cartersian Joins(Return everything)
```

```python
%%sql 
-- Generate all possible combinations of customers and orders


SELECT * 
FROM customers
CROSS JOIN orders

```

```python
# SET Opreators
#RULE 1:
#SET operator can be used almost in all clauses
#WHERE|JOIN|GROUP BY|HAVING
#ORDER BY is allowed only once at the end of the query

#RULE 2: The number of columns in each query must be the same

#RULE 3: Data types of columns in each query must be compatiable

#RULE 4: The order of the coulmns in each query must be the same

#RULE 5: The column names in the result set are determined by the column names specified in the first query

#RULE 6: MAPPING correct columns
```

```python
%%sql 
-- UNION (Returns all district rows from both queries and removes duplicate rows from the result)
-- Combines the data from employees and customers into one table

SELECT 
    first_name
FROM customers
UNION
SELECT 
    FirstName
FROM employees;


```

```python
%%sql
-- UNION ALL(Returns all rows from both queries, including duplicates)
-- UNION ALL is faster than UNION, you can check UNION ALL to identify duplicates
-- Combines the data from employees and customers into one table including duplicates

SELECT first_name 
from customers
UNION ALL 
SELECT FirstName
from employees

```

```python
%%sql 
-- EXCEPT(Returns all distinct rows from first query that are not found in the second query)
-- It is the only one where the order of queries affects the final result
-- Find employees who are not customers at the same time

SELECT 
    FirstName
FROM employees
EXCEPT
SELECT first_name 
from customers;

```

```python
%%sql 
-- INTERSECT(Returns only the rows that are common in both queries)
-- Find employees who are also customers
SELECT 
    FirstName
FROM employees
INTERSECT
SELECT first_name 
from customers;

```
