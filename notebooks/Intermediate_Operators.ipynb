```python
# Logical Operator
```

```python
%%sql
-- ADD Operator(All conditions must be TRUE)
-- Retrieve all customers who are from USA AND have score greater than 500

select *
from customers
where country = 'USA' and score > 500;
```

```python
%%sql
-- OR operator(Atleast one condition must be TRUE)
-- -- Retrieve all customers who are from USA OR have score greater than 500

select *
from customers
where country = 'USA' or score > 500;
```

```python
%%sql
-- NOT operator(Reverse) Exclides matching values
-- Retrive all customers with score NOT less than 500

select * 
from customers
where not score < 500


```

```python
# Range Operator
```

```python
%%sql 
-- Between(Checks if the value is in the range)
-- Retrieve all customers whose score falls in the range between 100 and 500

select * 
from customers
where score between 100 and 500

```

```python
# Membership Operator
```

```python
%%sql 
-- IN Operator (check if its in the list)
-- Retrieve all customers from either Germany or USA

select * 
from customers
where country in ('Germany','USA');
```

```python
%%sql 
-- IN Operator (check if its in the list)
-- Retrieve all customers from not in Germany or USA

select * 
from customers
where country not in ('Germany','USA');
```

```python
# Logical Operators
```

```python
%%sql 
-- LIKE - search for a pattern in text
-- % - Any number of character (0,1,many)
-- _ - Excat 1
-- Find all customers whose first name starts with 'M'

select * 
from customers
where first_name LIKE 'M%';
```

```python
%%sql
-- Find all customers whose first name ends with 'N'

select * 
from customers
where first_name LIKE '%N';
```

```python
%%sql
-- Find all customers whose first name contains 'r'

select * 
from customers
where first_name LIKE '%r%';
```

```python
%%sql
-- Find all customers whose first name has 'r' in the thirs position

select * 
from customers
where first_name LIKE '__r%';
```
