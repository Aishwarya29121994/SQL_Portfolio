```python
%%sql
INSERT INTO customers (id,first_name, country, score) 
VALUES (7,'Rishi', 'USA', NULL),
       (8,'Aishwarya', NULL, 100);
```

```python
%%sql 
-- change the score of customer with ID 6 ot 0

update customers
set score = 0
where id = 6 


```

```python
%%sql
-- change the score of customers with ID 10 to 0 and update the country to UK

update customers
set score = 0, 
    country = 'UK'
where id = 0 
```

```python
%%sql
-- Update all customers with a NULL score by setting their score to 0
update customers
set score = 0
where score is NULL
```
