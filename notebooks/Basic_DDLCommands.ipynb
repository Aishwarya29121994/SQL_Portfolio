```python
%load_ext sql
%sql sqlite:///portfolio.db
```

```python
%%sql
-- Create a new table called persons with columns : id, person_name, birth_date and phone

CREATE TABLE persons(
    id int not null,
    person_name varchar(50) not null,
    birth_date date,
    phone varchar(15) not null,
    constraint pk_persons primary key(id)
)
```

```python
%%sql
-- Add new column email to persons
-- Note : SQLite doesn’t allow you to add a new NOT NULL column without a default value
ALTER TABLE persons
ADD email VARCHAR(50) NOT NULL DEFAULT 'unknown@example.com';
```

```python
%%sql
select * from persons;
```

```python
%%sql
-- Remove phone from persons
alter table persons
drop column phone 
```

```python
%%sql
-- delete table persons from databases
drop table persons;
```
